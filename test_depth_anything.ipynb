```python
import sys, io, torch
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
from depth_anything_3.api import DepthAnything3

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using: {device}")

# Options: da3-small, da3-base, da3-large
model = DepthAnything3.from_pretrained("depth-anything/da3-base")
model = model.to(device=device)
```

```python
import io
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
from google.colab import files

# Upload
uploaded = files.upload()
IMAGE_PATH = list(uploaded.keys())[0]
with open(IMAGE_PATH, "wb") as f:
    f.write(uploaded[IMAGE_PATH])
image = Image.open(io.BytesIO(uploaded[IMAGE_PATH])).convert("RGB")

# Inference
prediction = model.inference([IMAGE_PATH], export_dir="output", export_format="npz")
depth = prediction.depth[0]
print(f"depth: {depth.shape}  min={depth.min():.3f}  max={depth.max():.3f}")

# Visualize
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
axes[0].imshow(image)
axes[0].set_title("Input")
axes[0].axis("off")
im = axes[1].imshow(depth, cmap="inferno")
axes[1].set_title("Depth")
axes[1].axis("off")
plt.colorbar(im, ax=axes[1], fraction=0.046, pad=0.04)
plt.tight_layout()
plt.show()
```
